In this notebook we check the dynamics of spin observables for the BTC.

```python
import numpy as np
import btc
from btc_fixed import run_collective_trajectories
from btc_fixed import run_quantum
from pathlib import Path
import datetime
import matplotlib.pyplot as plt
```

```python
cspins = run_collective_trajectories(10, 20, 0, 0., np.linspace(0, 10, 10000),
                                                   n_traj=500, seed=42, scheme="diffusive")
```

```python
(qspins, times) = run_quantum(5, 20, 0., np.linspace(0, 10, 1000)) , np.linspace(0, 10, 1000)
```

```python
plt.figure(figsize=(10, 6))
plt.plot(cspins["times"], np.mean(cspins["mx"], axis=1), label='mx (classical)', color='blue')
plt.plot(cspins["times"], np.mean(cspins["my"], axis=1), label='my (classical)', color='orange')
plt.plot(cspins["times"], np.mean(cspins["mz"], axis=1), label='mz (classical)', color='green')
plt.plot(times, qspins["sx"], label='qspin 1 (quantum)', color='red', linestyle='--')
plt.plot(times, qspins["sy"], label='qspin 2 (quantum)', color='purple', linestyle='--')
plt.plot(times, qspins["sz"], label='qspin 3 (quantum)', color='brown', linestyle='--')
plt.xlabel('Time')
plt.ylabel('Magnetization') 
```
